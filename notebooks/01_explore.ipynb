```python
import pandas as pd

df = pd.read_csv("../data/raw/covid19br.csv")
print(df.shape)
print(df.columns.tolist())
print(df.head())
print(df.isna().sum())
```

```python
print(df.dtypes)
print(df[['midia','url','viral','sharings','characters','words']].describe())
print(df['viral'].value_counts())
print(df['midia'].value_counts())
print(df['sharings'].value_counts().head(10))
print("groups:", df['group'].nunique(), " states:", df['state'].nunique())
dates = pd.to_datetime(df['date'], format='%d/%m/%y')
print(dates.min(), dates.max())
```

```python
print("unique texts:", df['text'].nunique(), "of", len(df))
print(df['text'].value_counts().head(15))
```

```python
long = df[df['words'] >= 15]
print("messages with 15+ words:", len(long))
for t in long['text'].sample(5, random_state=1):
    print(t)
    print("-----")
```
